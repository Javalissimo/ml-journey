```python
def read_amount(raw: str) -> float | None:
    try:
        value = float(raw)
    except ValueError:
        print('не число:', raw)
        return None
    else:
        print('ok')
        return value
    finally:
        print('проверено')

print(read_amount('12.5'))
print(read_amount('abc'))
```
