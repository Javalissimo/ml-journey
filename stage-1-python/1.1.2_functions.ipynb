```python
def make_bonus_row(client):
    client_with_bonus = {'name': client['name'], 'bonus_x2': client['bonus'] * 2}
    return client_with_bonus

clients = [
    {"id": 1, "name": "Анна", "city": "Казань", "bonus": 120},
    {"id": 2, "name": "Борис", "city": None, "bonus": 0},
    {"id": 3, "name": "Вера", "city": "Сочи", "bonus": 50},
]

new_client = [make_bonus_row(client) for client in clients]
print(new_client)
```

```python
def normalize_city(city: str) -> str:
    """Приводит название города к виду 'Казань'."""
    if city is None:
        return 
    return city.strip().capitalize()

print(normalize_city(None))
print(normalize_city("казань"))
print(normalize_city("  СОЧИ "))
```

```python
def price_with_discount(price, percent=10):
    return price * (1 - percent / 100)
print(price_with_discount(100))
print(price_with_discount(100,20))
print(price_with_discount(100, percent = 0))
```
