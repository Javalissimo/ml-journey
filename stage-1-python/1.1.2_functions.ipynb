```python
def make_bonus_row(client):
    client_with_bonus = {'name': client['name'], 'bonus_x2': client['bonus'] * 2}
    return client_with_bonus

clients = [
    {"id": 1, "name": "Анна", "city": "Казань", "bonus": 120},
    {"id": 2, "name": "Борис", "city": None, "bonus": 0},
    {"id": 3, "name": "Вера", "city": "Сочи", "bonus": 50},
]

new_client = [make_bonus_row(client) for client in clients]
print(new_client)
```

```python
def normalize_city(city: str) -> str:
    """Приводит название города к виду 'Казань'."""
    if city is None:
        return 
    return city.strip().capitalize()

print(normalize_city(None))
print(normalize_city("казань"))
print(normalize_city("  СОЧИ "))
```

```python
def price_with_discount(price, percent=10):
    return price * (1 - percent / 100)
print(price_with_discount(100))
print(price_with_discount(100,20))
print(price_with_discount(100, percent = 0))
```

```python
def avg_check(*totals: float) -> float:
    if not totals:
        return 0.0
    print(type(totals))   
    return sum(totals) / len(totals)

print(avg_check(500, 700, 300))
print(avg_check())
```

```python
orders = [
    {"order_id": 1, "city": "Казань", "total": 1500},
    {"order_id": 2, "city": "Москва", "total": 300},
    {"order_id": 3, "city": "Казань", "total": 900},
    {"order_id": 4, "city": "Сочи",   "total": 2100},
]

sorted_orders = sorted(orders, key=lambda total: total['total'], reverse=True)
print(sorted_orders)
max_order_id = max(orders, key = lambda o: o['total'])
print(max_order_id['order_id'])
```

```python
def add_vat(order: dict) -> dict:
    """Чистая: возвращает новый словарь с НДС, оригинал не трогает."""
    return {**order, "total": order["total"] * 1.2}

order = {"order_id": 1, "total": 1000}   # создаём заказ заново
new_order = add_vat(order)
print(order)                  # оригинал
print(new_order)              # результат
print(order is new_order)     # один и тот же объект?
```

```python
def add_item(item: str, items: list = []) -> list:
    """Добавляет товар в корзину и возвращает корзину."""
    items.append(item)
    return items

cart_1 = add_item("хлеб")
cart_2 = add_item("молоко")
print(cart_1)
print(cart_2)
print(cart_1 is cart_2)
```
