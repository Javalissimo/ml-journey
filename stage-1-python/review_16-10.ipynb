```python
import pprint
orders = [
    {"order_id": 1, "city": "Казань", "total": 1500},
    {"order_id": 2, "city": "Москва", "total": 300},
    {"order_id": 3, "city": "Казань", "total": 900},
    {"order_id": 4, "city": "Сочи",   "total": 2100},
]

def apply_discount(order: dict, percent: float ) -> dict:    
    per = 1 - percent / 100    
    return [{"order_id":o['order_id'], "city":o['city'], "total":o['total']*per} for o in order]

pprint.pprint(apply_discount(orders, 10))

pprint.pprint(orders)


```

```python
import pprint
orders = [
    {"order_id": 1, "city": "Казань", "total": 1500},
    {"order_id": 2, "city": "Москва", "total": 300},
    {"order_id": 3, "city": "Казань", "total": 900},
    {"order_id": 4, "city": "Сочи",   "total": 2100},
]

min_order = min(orders, key=lambda o: o['total'])
print(min_order['order_id'])

sorted_order = sorted(orders, key=lambda o: o['city'])

print([o['order_id'] for o in sorted_order])
```

```python
def add_vat(order: dict) -> dict:
    new_order = order.copy()
    new_order['total'] *= 1.2
    return new_order

new_order = [add_vat(o) for o in orders]
pprint.pprint(new_order)
```
