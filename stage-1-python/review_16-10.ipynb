```python
import pprint
orders = [
    {"order_id": 1, "city": "Казань", "total": 1500},
    {"order_id": 2, "city": "Москва", "total": 300},
    {"order_id": 3, "city": "Казань", "total": 900},
    {"order_id": 4, "city": "Сочи",   "total": 2100},
]

def apply_discount(order: dict, percent: float ) -> dict:    
    per = 1 - percent / 100    
    return [{"order_id":o['order_id'], "city":o['city'], "total":o['total']*per} for o in order]

pprint.pprint(apply_discount(orders, 10))

pprint.pprint(orders)


```
