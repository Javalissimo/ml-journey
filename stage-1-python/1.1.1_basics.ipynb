```python
sep = {"Анна", "Борис", "Вера"}     # купили в сентябре
oct = {"Борис", "Вера", "Глеб"}     # купили в октябре

#sep & oct    # {'Борис', 'Вера'}            — пересечение: купили в ОБА месяца
#sep | oct    # {'Анна', 'Борис', 'Вера', 'Глеб'} — объединение: все клиенты
sep - oct    # {'Анна'}           
```
