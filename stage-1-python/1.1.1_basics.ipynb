price_list = {"молоко": 89.9, "хлеб": 45.0, "сыр": 699.0, "кефир": 120.5, "масло": 259.0}

## Задача 3

orders = [
    ("A-001", "Москва", 1200),
    ("A-002", "Сочи", 800),
    ("A-003", "москва", 500),
    ("A-004", "Казань", 300),
    ("A-005", "Сочи ", 1500),
    ("A-006", "Казань", 2100),
]

Сделать:

а) множество уникальных очищенных городов (должно получиться 3, а не 5);
б) словарь выручки по очищенным городам (подумай, что здесь подходит: comprehension или цикл);
в) одной строкой — список номеров заказов на сумму ≥ 1000;
г) город с максимальной выручкой (можно циклом).

```python
orders = [
    ("A-001", "Москва", 1200),
    ("A-002", "Сочи", 800),
    ("A-003", "москва", 500),
    ("A-004", "Казань", 300),
    ("A-005", "Сочи ", 1500),
    ("A-006", "Казань", 2100),
]

orders_set = {city.strip().lower() for number, city, price in orders}
print(orders_set)

revenue = {}
for number, city, price in orders:
    clean_city = city.strip().lower()
    if clean_city in revenue:
        revenue[clean_city] += price
    else:
        revenue[clean_city] = price
print(revenue)

orders_list = [number for number,_,price in orders if price >= 1000]
print(orders_list)

max_city = None
max_price = 0
for city, price in revenue.items():
    if price > max_price:
        max_city = city
        max_price = price

print(f"Город с максимальной выручкой:{max_city}-{max_price}")
```

```python
receipts = [["хлеб", "молоко"], ["сыр"], ["хлеб", "кефир", "молоко"], ["хлеб"]]

# 1. Первые два чека — для проверки глазами
first_two = receipts[:2]

# 2. Сколько раз встречается каждый товар
counts = {}
for receipt in receipts:
    for item in receipt:
        counts[item] = counts.get(item, 0) + 1

# 3. Товары, которые купили больше одного раза
#    (for item in counts перебирает ключи словаря)
popular = [item for item in counts if counts[item] > 1] # Ошибка - конструкция if else должна быть сначала, но тут лучше просто испоьзовать if

# 4. Сохраняем исходные подсчёты в backup, а в рабочей версии обнуляем хлеб
backup = counts.copy()
counts["хлеб"] = 0

print(first_two)        # ожидаем: [['хлеб', 'молоко'], ['сыр']]
print(popular)          # ожидаем: ['хлеб', 'молоко']
print(backup["хлеб"])   # ожидаем: 3
```

## Задача 5

Дано: поля клиента в старой и новой системе:

old_fields = ["client_id", "Full_Name", "phone", "email", "birth_date", "city"]
new_fields = ["client_id", "full_name", "email", "phone_number", "city", "created_at"]

record = {"client_id": 17, "Full_Name": "Иван Петров", "phone": "+79990000000",
          "email": "ivan@mail.ru", "birth_date": "1990-05-12", "city": "Казань"}

а) не обращая внимания на регистр, найди поля, которые есть в обеих системах, только в старой и только в новой;
б) составь словарь-маппинг {поле_старой: поле_новой}: автоматические совпадения плюс те, что очевидно соответствуют друг другу по смыслу;
в) одной строкой переименуй ключи record по маппингу. Поля без маппинга не переносятся;
г) 2–3 предложениями: какие поля потеряются при переносе и что ты как аналитик сказал бы об этом заказчику.

```python
old_fields = ["client_id", "Full_Name", "phone", "email", "birth_date", "city"]
new_fields = ["client_id", "full_name", "email", "phone_number", "city", "created_at"]

record = {"client_id": 17, "Full_Name": "Иван Петров", "phone": "+79990000000",
          "email": "ivan@mail.ru", "birth_date": "1990-05-12", "city": "Казань"}

old_set = {value.lower() for value in old_fields}
new_set = {value.lower() for value in new_fields}
print(f'поля в обеих системах {old_set & new_set}')
print(f'поля только старой системы {old_set - new_set}')
print(f'поля только новой системы {new_set - old_set}')

mapping = {'client_id': 'client_id',
           'Full_Name': 'full_name',
           'phone': 'phone_number',
           'email': 'email',            
           'city': 'city'}

new_record = {mapping[old_key]: value for old_key, value in record.items() if old_key in mapping}
print(new_record)
```
