```python
orders = [
    {'order_id': 101, 'city': 'Казань', 'total': 1200},
    {'order_id': 102, 'city': 'Москва', 'total': 450},
    {'order_id': 103, 'city': 'Казань', 'total': 3100},
    {'order_id': 104, 'city': 'Самара', 'total': 800},
]

revenue = {'Казань': 4300, 'Москва': 450, 'Самара': 800}

comprehension_orders = [o['order_id'] for o in orders if o['total'] > 1000]
print(comprehension_orders)
print(max(orders, key=lambda o:o['total']))
for o,v in revenue.items():
    print(f'{o}: {v}')

import sys                 # модуль с информацией о запущенном Python
print(sys.executable)      # путь к python.exe, который сейчас выполняет ноутбук
```
