```python
def revenue_by_city(orders, status=None) -> dict[str, int]:
    """возвращает словарь {город: сумма total};
если status передан, учитываются только заказы с этим статусом;
если status=None, учитываются все заказы;
"""
    result = {}
    
    for o in orders:
        city = o['city']
        if status is not None:
            if o['status'] == status:
                result[city] = result.get(city, 0) + o['total']
        else:
            result[city] = result.get(city, 0) + o['total']        
    return result

def print_city_revenue(order,status=None):
    result_order = sorted(revenue_by_city(order,status).items(), key=lambda o: o[1],reverse=True)
    for city, total in result_order:
        print(f'{city}: {total}')


print_city_revenue(orders)
print_city_revenue(orders, status='доставлен')
print_city_revenue(orders, status='потерян')
```

```python
TAX = 0.2
orders = [
    {'order_id': 101, 'city': 'Казань', 'total': 1200},
    {'order_id': 102, 'city': 'Москва', 'total': 450},
    {'order_id': 103, 'city': 'Казань', 'total': 3100},
]

def get_total(order):
    return order['total']

def most_expensive(orders):
    return max(orders, key=get_total)

def add_item(item, items=None):
    if items is None:
        items = []  
    items.append(item)
    return items

def add_tax(order):
    new_order = order.copy()  
    new_order['total'] = new_order['total'] * (1 + TAX)
    return new_order

taxed = add_tax(orders[0])
print(orders[0])

print(orders)
```
