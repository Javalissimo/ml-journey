```python
from pathlib import Path
DATA_DIR = Path.cwd().parent / 'data'
raw_path = DATA_DIR / 'orders_raw.csv'
text = """order_id,city,amount
101, казань ,1 200
102,Москва,800
103,,abc
104,санкт-петербург,2 500
"""
with open(raw_path, 'w', encoding='utf-8') as f:
    f.write(text)
print(raw_path.exists())
```

```python
import csv
with open(raw_path, encoding='utf-8', newline='') as f:
    rows = list(csv.DictReader(f))

for row in rows:
    print(row, type(row['amount']))
```

```python
from orders_utils import normalize_city, parse_amount


clean_rows =[]
for row in rows:
    temp_dict={'order_id':row['order_id'], 'city':normalize_city(row['city']), 'amount':parse_amount(row['amount'])}    
    clean_rows.append(temp_dict)

#new_clean_rows = [{'order_id':row['order_id'], 'city':normalize_city(row['city']), 'amount':parse_amount(row['amount'])} for row in rows]


import csv

clean_path = DATA_DIR / 'orders_clean.csv'
print(clean_path)

with open(clean_path, 'w', encoding='utf-8', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=['order_id', 'city', 'amount'])
    writer.writeheader()
    writer.writerows(clean_rows)
```
