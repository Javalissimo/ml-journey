```python
from sklearn.tree import DecisionTreeClassifier

X = [[5], [12], [30], [45], [70], [90], [120], [150]]  # признак: дней с последнего заказа
y = [0, 0, 0, 0, 1, 1, 1, 1]                            # таргет: 1 = клиент ушёл

model = DecisionTreeClassifier(max_depth=1).fit(X, y)
print("Порог, найденный моделью:", model.tree_.threshold[0])
print("Прогноз для 50 и 100 дней:", model.predict([[50], [100]]))
```
