```python
from pathlib import Path    
import sqlite3              
import pandas as pd

DB_PATH = Path("../data/shop.db")   
print(DB_PATH.exists())             

conn = sqlite3.connect(DB_PATH)     

query = """
SELECT full_name, city
FROM customers
WHERE city <> 'Казань' OR city IS NULL
ORDER BY city;
"""                                 

df = pd.read_sql(query, conn)       
conn.close()                        

print(len(df))                      
df.head(10)                         
```
